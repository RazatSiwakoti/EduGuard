```python
from pathlib import Path

import joblib
import pandas as pd


MODEL_PATH = Path(__file__).with_name("student_risk_model.joblib")

artifacts = joblib.load(MODEL_PATH)

model = artifacts["model"]
feature_names = artifacts["feature_names"]
label_encoder = artifacts["label_encoder"]

print("Features:", feature_names)
print("Classes:", list(label_encoder.classes_))

sample_values = {
    "attendance_pct": 72.0,
    "attendance_trend": -5.0,
    "tut_completion_pct": 80.0,
    "tut_trend": 0.0,
    "assessment_avg_pct": 68.0,
    "moodle_login_count": 12,
}

sample = pd.DataFrame([sample_values]).reindex(columns=feature_names)

prediction_index = model.predict(sample)[0]
probabilities = model.predict_proba(sample)[0]
prediction_label = label_encoder.inverse_transform(
    [int(prediction_index)]
)[0]

print("Prediction:", prediction_label)
print("Probabilities:", probabilities)
```
